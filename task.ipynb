```python
table_turns = {
    "T1":5,
    "T2":2,
    "T3":4,
    "T4":1,
    "T5":3,
    "T6":6,
    "T7":2,
    "T8":4}

high=[]
average=[]
low=[]

for table, turns in table_turns.items():
    
    if turns >=4:
        high.append(table)

    elif turns>=2:
        average.append(table)
        
    else:
        low.append(table)

total = sum(table_turns.values())

print("High Performing:", high, len(high),"tables")
print("Average:", average, len(average), "table")
print("low:", low, len(low), "table")
print("Total Groups Served:", total)
      
```
